```python
import pandas as pd, numpy as np
vc = pd.read_csv("ai_startup_funding_sample.csv")
vc.head()
```

```python
print(vc["country"].value_counts().to_string())
```

```python
# Tidy the spaces and capitals
vc["country"] = vc["country"].str.strip().str.title()

# Swap the leftover name for the standard one
vc["country"] = vc["country"].replace({'England, United Kingdom': 'United Kingdom', 'United States (Usa)': 'United States', 'Ger': 'Germany', 'Can': 'Canada'})

print(vc["country"].value_counts().to_string())
```

```python
#check for empty values
vc.isna().sum()
```

```python
#check which rows hv empty values
vc[vc["country"].isna()]
```

```python
#assign rows w missing values to country based on cities
#dropped one value (Dresma) as it is missing both country + city
vc.loc[vc["round_id"].isin(["11e2ae23ac99", "3c0ed93fba03"]), "country"] = "United States"
vc.loc[vc["round_id"].isin(["63017d017fca"]), "country"] = "Canada"
vc.isna().sum()
```

```python
#check for duplicates
vc.duplicated().sum()
```

```python
#reduce to us, uk, germany and france + timeframe from 2021-2025 + verified status
vc= vc[vc["country"].isin(["United States", "France", "United Kingdom", "Germany"]) & vc["year"].between(2021, 2025)]
print(len(vc))

#filters depending on country within country column
uk = vc[vc["country"] == "United Kingdom"]
us = vc[vc["country"] == "United States"]
fr = vc[vc["country"] == "France"]
de = vc[vc["country"] == "Germany"]
```

```python
#top 5 highest in the uk for manual checking
uk.nlargest(5, "amount_usd")
```

```python
#remove nscale bcs it raised the amt in 2026 not 2023
uk = uk[uk["round_id"] != "26a87e5fa28e"]
```

```python
#Remove Nscale from the main vc dataframe (not just UK)
vc = vc[vc["round_id"] !="26a87e5fa28e"].copy()
```

```python
#top 5 highest in the US to manual check
us.nlargest(5, "amount_usd")
```

```python
#repeat for germany
de.nlargest(5, "amount_usd")
```

```python
#repeat for france
fr.nlargest(5, "amount_usd")
```

```python
#new columns for easier classifications
vc["period"] = np.where(vc["year"] <= 2022, "2021-22", "2023-25")

def stage_group(s):
    if s in ["Pre-seed", "Seed", "Angel", "Pre-Series A"]: return "Early"
    if s in ["Series A", "Series B", "Pre-Series B"]: return "Series A-B"
    if s in ["Series C", "Series D", "Series E", "Series F", "Series G", "Series H", "Growth / PE"]: return "Late"
    if s == "Undisclosed": return "Undisclosed"
    return "Other"
vc["stage_group"] = vc["stage"].map(stage_group)
```

```python
#1. Total number of deals per country
vc.groupby(["country", "period"])["round_id"].count()
```

```python
#2. Median Deal Size Per Country (Before vs After)
vc.groupby(["country", "period"])["amount_usd"].median()
```

```python
#3. UK Stage Breakdown
uk = vc[vc["country"] == "United Kingdom"]
uk.groupby(["period", "stage_group"])["round_id"].count()
```

OUR DECISION LOG:


*   Standardised country names
*   Dropped 1 row with both country and city values missing (Dresma)
*   Assigned missing country values for 3 other rows based on their city data
*   Checked for duplicates
*   Dropped years outside of the 2021-2025 window
*   Dropped non-target countries other than United States, United Kingdom, France and Germany
*   Dropped outlier/misdated value- Nscale was logged under 2023 but manually checked to be 2026
*   Created a period column to split data into pre-ChatGPT and post-ChatGPT timeline
*   Grouped funding rounds into broader stage categories
